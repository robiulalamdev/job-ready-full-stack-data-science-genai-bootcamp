```python
import pandas as pd
import seaborn as sns
```

```python
# Load Titanic CSV Data
titanic = pd.read_csv("./train.csv")
```

```python
titanic.head()
```

# Load tips data from seaborn

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

# Load Flights data from seaborn

```python
flights = sns.load_dataset("flights")
```

```python
flights.head()
```

# Load Iris data from seaborn

```python
iris = sns.load_dataset("iris")
```

```python
iris.head()
```

# 1. Scatterplot (Numerical - Numerical)

```python
tips.sample(5)
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue=tips["sex"], style=tips["smoker"])
```

## 2. Bar Plot (Numerical - Categorical)

```python
titanic.head()
```

```python
sns.barplot(data=titanic, x='Pclass', y="Age", hue=titanic["Sex"])
```

## 3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(titanic, x=titanic["Pclass"], y=titanic["Age"], hue=titanic["Survived"])
```

## 4. Distplot (Numerical - Categorical)

```python
sns.distplot(titanic[titanic["Survived"]==0]["Age"],  hist=False)
sns.distplot(titanic[titanic["Survived"]==1]["Age"], hist=False)
```
