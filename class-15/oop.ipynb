```python
class Person:
    def __init__(self): # Dunder/magic method
        self.name = "Robiul"
        self.age = 23
        self.address = "Thakurgaon Bangladesh"

    # Class Properties
    def getInformation(self):
        print(f'Name: {self.name} Age: {self.age} Address: {self.address}')

    def changeName(self, name:str):
        newName = str(name)
        self.name= newName

    def changeAge(self, age:int):
        newAge = int(age)
        self.age = newAge

    def changeAddress(self, address:str):
        newAddress = str(address)
        self.address = newAddress
```

```python
robiul = Person()
```

```python
name = robiul.name
print(name)
```

```python
age = robiul.age
print(age)
```

```python
address = robiul.address
print(address)
```

```python
robiul.changeName("Sakil Khan")
robiul.changeAge(21)
robiul.changeAddress("Boguladangi, bhully bazar, thakurgaon")

print(robiul.name, robiul.age, robiul.address)
```
