# Data Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from numpy.random import randn, randint, sample, uniform
```

```python
# import sample data
iris = sns.load_dataset("iris")
```

```python
iris.head()
```

```python
ax = iris.plot(figsize=(15,8), title="Irish Dataset")
ax.set_xlabel="X axis"
ax.set_ylabel="Y axis"
```

```python
ldf = iris.drop(["species"], axis=1)
```

```python
ldf.head()
```

```python
ldf.plot(figsize=(15,8), title="Dataset")
```

```python
ldf.iloc[0]
```

```python
ldf.iloc[0].plot(kind="bar")
```
