# Modify & Clean (2 Min)

```python
# Add new column
data["Total"]= data["Recommends"]*2.5
```

```python
data.head()
```

```python
# Drop a column or row
data = data.drop(columns=['Total'])
```

```python
# get missing data
data.isna().sum()
```

```python
# Drop rows with missing values
data = data.dropna()
```

# Group & Aggregate (2 Min)

```python
data.groupby("CollTitle")["order"].mean()
data.groupby("CollTitle").agg({
    'order': ["mean", "max"],
    'Recommends':'min'
})
```
